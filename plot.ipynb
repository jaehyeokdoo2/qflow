```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import json
import os
```

```python
def load_json(path):
    with open(path, "r") as f:
        return json.load(f)
```

```python
# Only for manipulation tasks
def load_qflow_results_csv(env_name, alpha, flow_steps, q_agg, discount, start_seed, end_seed, time_embed_dim, actor_loss_type="grad", metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/rebuttal/"
    results = []

    if time_embed_dim is not None:
        time_embed_str = f'te{time_embed_dim}'
    else:
        time_embed_str = ''
    tmp_env_name = env_name.replace("-v0", "").replace("-singletask-task", '-t')
    
    init_seed = 10001
    for seed in range(start_seed, end_seed + 1):
        # seed = init_seed + init_seed * seed
        exp_name = f"qflow-{tmp_env_name}-a{alpha}-qagg{q_agg}-{time_embed_str}-al{actor_loss_type}-sd{seed}"
        eval_results_dir = os.path.join(root_dir, "qflow", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)


def load_fql_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/fql-reproduce/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"fql_{env_name}_sd{seed}"
        eval_results_dir = os.path.join(root_dir, "fql", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

def load_fbrac_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/fbrac-reproduce/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"fbrac_{env_name}_sd{seed}"
        eval_results_dir = os.path.join(root_dir, "fbrac", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

def load_qipo_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/qipo-reproduce/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"qipo_{env_name}_sd{seed}"
        eval_results_dir = os.path.join(root_dir, "qipo", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

def load_ifql_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/ifql-reproduce/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"ifql_{env_name}_sd{seed}"
        eval_results_dir = os.path.join(root_dir, "ifql", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)
```

```python
def load_qflow_ogbnech_results_csv(env_name, alpha, q_agg, discount, start_seed, end_seed, time_embed_dim, actor_loss_type="grad", metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/rebuttal/"
    results = []

    if time_embed_dim is not None:
        time_embed_str = f'te{time_embed_dim}'
    else:
        time_embed_str = ''
    tmp_env_name = env_name.replace("-v0", "").replace("-singletask-task", '-t')
    
    init_seed = 10001
    for seed in range(start_seed, end_seed + 1):
        # seed = init_seed + init_seed * seed
        exp_name = f"qflow-{tmp_env_name}-a{alpha}-qagg{q_agg}-{time_embed_str}-al{actor_loss_type}-sd{seed}"
        eval_results_dir = os.path.join(root_dir, "qflow", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

hyperparams_dict = {
    "antmaze_large":{
        "alpha":1,
        "q_agg":"mean",
        "discount":0.99,
    },
    "antmaze_giant":{
        "alpha":0.2,
        "q_agg":"min",
        "discount":0.995,
    },
    "humanoid_medium":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.995,
    },
    "humanoid_large":{
        "alpha":1,
        "q_agg":"mean",
        "discount":0.995,
    },
    "antsoccer":{
        "alpha":0.5,
        "q_agg":"mean",
        "discount":0.995,
    },
    "cube_single":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.99,
    },
    "cube_double":{
        "alpha":10,
        "q_agg":"mean",
        "discount":0.99,
    },
    "puzzle_3":{
        "alpha":20,
        "q_agg":"min",
        "discount":0.99,
    },
    "puzzle_4":{
        "alpha":100,
        "q_agg":"min",
        "discount":0.99,
    },
    "scene":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.99,
    },   
}

all_results = []

for task_idx in range(4, 5):
    env_names = {
        "antmaze_large": f"antmaze-large-navigate-singletask-task{task_idx}-v0",
        "antmaze_giant": f"antmaze-giant-navigate-singletask-task{task_idx}-v0",
        "humanoid_medium": f"humanoidmaze-medium-navigate-singletask-task{task_idx}-v0",
        "humanoid_large": f"humanoidmaze-large-navigate-singletask-task{task_idx}-v0",
        "antsoccer": f"antsoccer-arena-navigate-singletask-task{task_idx}-v0",
        "cube_single": f"cube-single-play-singletask-task{task_idx}-v0",
        "cube_double": f"cube-double-play-singletask-task{task_idx}-v0",
        "puzzle_3": f"puzzle-3x3-play-singletask-task{task_idx}-v0",
        "puzzle_4": f"puzzle-4x4-play-singletask-task{task_idx}-v0",
        "scene": f"scene-play-singletask-task{task_idx}-v0",
        # ... other envs
    }
    env_key = 'puzzle_4'
    env_name = env_names[env_key]

    results = load_qflow_ogbnech_results_csv(
        env_name=env_name,
        start_seed=0,
        end_seed=3,
        time_embed_dim=16,
        actor_loss_type='grad',
        **hyperparams_dict[env_key])

    all_results.append(results)  # each: (num_seeds, num_timesteps)

# Concatenate across tasks: (num_tasks * num_seeds, num_timesteps)
all_results = np.concatenate(all_results, axis=0).reshape(1, 4, -1)

mean = all_results.mean(axis=0)
print(mean)
final3_per_seed = mean[:, -3:].mean(axis=0)
final1_per_seed = mean[:, -1]

print(f"Final 3 averages: {final3_per_seed.mean():.2f} ± {final3_per_seed.std():.2f}")
print(f"Final only average: {final1_per_seed.mean():.2f} ± {final1_per_seed.std():.2f}")
```

```python
def load_qflow_online_results_csv(env_name, alpha, q_agg, time_embed_dim, actor_loss_type="grad", start_seed=0, end_seed=4, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/online_finetuning/"
    results = []

    if time_embed_dim is not None:
        time_embed_str = f'te{time_embed_dim}'
    else:
        time_embed_str = ''

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"qflow-{env_name}-a{alpha}-qagg{q_agg}-{time_embed_str}-al{actor_loss_type}-sd{seed}_online"
        eval_results_dir = os.path.join(root_dir, "qflow", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

def load_fql_online_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/online_finetuning/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"fql_{env_name}_sd{seed}_online"
        eval_results_dir = os.path.join(root_dir, "fql", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)


def load_fbrac_online_results_csv(env_name, start_seed, end_seed, metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/online_finetuning/"
    results = []

    for seed in range(start_seed, end_seed + 1):
        exp_name = f"fbrac_{env_name}_sd{seed}_online"
        eval_results_dir = os.path.join(root_dir, "fbrac", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)
```
