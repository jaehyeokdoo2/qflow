```python
# ── Bootstrap CI ──────────────────────────────────────────────────────
def bootstrap_ci(data, n_boot=2000, ci=95, seed=0):
    """data: (n_seeds, n_steps) → lo, hi each of shape (n_steps,)"""
    rng = np.random.default_rng(seed)
    idx = rng.integers(0, len(data), size=(n_boot, len(data)))
    boots = data[idx].mean(axis=1)
    lo = np.percentile(boots, (100 - ci) / 2, axis=0)
    hi = np.percentile(boots, 100 - (100 - ci) / 2, axis=0)
    return lo, hi

def _load_antmaze(root_dir, method, env_name, exp_name_fn, start_seed, end_seed, metric_name):
    curves = []
    for seed in range(start_seed, end_seed + 1):
        path = os.path.join(root_dir, method, env_name, exp_name_fn(seed), "eval.csv")
        if os.path.exists(path):
            df = pd.read_csv(path)
            curves.append(df[f"evaluation/{metric_name}"].values)
        else:
            print(f"Missing: {path}")
    if not curves:
        return None, np.array([])
    max_len = max(len(c) for c in curves)
    completed = [c for c in curves if len(c) == max_len]
    if len(completed) < len(curves):
        print(f"Skipping {len(curves) - len(completed)} incomplete run(s)")
    # Each eval is logged every 100K training steps
    steps = np.arange(0, max_len) * 100_000
    return steps, np.array(completed)


def load_qflow_antmaze(env_name, alpha, start_seed, end_seed, metric_name="episode.normalized_return"):
    root_dir = "/mnt/nas/jaehyeok/qflow/qflow-reproduce/"
    return _load_antmaze(root_dir, "qflow", env_name,
                         lambda sd: f"qflow-{env_name}-a{alpha}-qaggmean-te16-algrad-sd{sd}",
                         start_seed, end_seed, metric_name)

def load_fql_antmaze(env_name, start_seed, end_seed, metric_name="episode.normalized_return"):
    root_dir = "/mnt/nas/jaehyeok/qflow/fql-reproduce/"
    return _load_antmaze(root_dir, "fql", env_name,
                         lambda sd: f"fql_{env_name}_sd{sd}",
                         start_seed, end_seed, metric_name)

def load_fbrac_antmaze(env_name, start_seed, end_seed, metric_name="episode.normalized_return"):
    root_dir = "/mnt/nas/jaehyeok/qflow/fbrac-reproduce/"
    return _load_antmaze(root_dir, "fbrac", env_name,
                         lambda sd: f"fbrac_{env_name}_sd{sd}",
                         start_seed, end_seed, metric_name)

# ── Config ────────────────────────────────────────────────────────────
ENV_KEYS = ["umaze", "umaze_diverse", "medium_play", "medium_diverse", "large_play", "large_diverse"]
env_names = {
    "umaze":          "antmaze-umaze-v2",
    "umaze_diverse":  "antmaze-umaze-diverse-v2",
    "medium_play":    "antmaze-medium-play-v2",
    "medium_diverse": "antmaze-medium-diverse-v2",
    "large_play":     "antmaze-large-play-v2",
    "large_diverse":  "antmaze-large-diverse-v2",
}
SHORT_NAMES = {
    "umaze":          "Umaze-default",
    "umaze_diverse":  "Umaze-diverse",
    "medium_play":    "Medium-play",
    "medium_diverse": "Medium-diverse",
    "large_play":     "Large-play",
    "large_diverse":  "Large-diverse",
}
method_order  = ["QFlow"]
method_colors = {"QFlow": "#2188DC", "FQL": "#91D391", "FBRAC": "#CD90CD"}

# ── Load ──────────────────────────────────────────────────────────────
all_data = {}  # {env_key: {method: (steps, curves)}}
for env_key in ENV_KEYS:
    env_name   = env_names[env_key]
    alpha      = hyperparams_dict[env_key]["alpha"]
    metric     = hyperparams_dict[env_key]["metric_name"]
    all_data[env_key] = {
        "QFlow": load_qflow_antmaze(env_name, alpha=alpha, start_seed=0, end_seed=7, metric_name=metric),
    }

# ── Plot ──────────────────────────────────────────────────────────────
N_ENVS = len(ENV_KEYS)
fig, axes = plt.subplots(1, N_ENVS, figsize=(N_ENVS * 4, 3), dpi=200)

for col, env_key in enumerate(ENV_KEYS):
    ax = axes[col]
    for method in method_order:
        steps, curves = all_data[env_key][method]
        if curves.size == 0:
            continue
        steps_m = steps / 1e6
        mean    = curves.mean(axis=0)
        lo, hi  = bootstrap_ci(curves)
        ax.plot(steps_m, mean, label=method, color=method_colors[method], linewidth=2.0, marker='o')
        ax.fill_between(steps_m, lo, hi, color=method_colors[method], alpha=0.15)

    ax.set_title(SHORT_NAMES[env_key], fontsize=12)
    ax.set_ylim(0, 105)
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(color="gray", linestyle=":", linewidth=0.4, alpha=0.5)
    ax.set_axisbelow(True)
    ax.tick_params(labelsize=10)
    ax.set_facecolor("#fcfcfc")
    ax.set_xlabel("Steps (M)", fontsize=12)
    if col == 0:
        ax.set_ylabel("Score", fontsize=12)

plt.tight_layout()
plt.savefig("d4rl_antmaze_training_curves.pdf", dpi=300)
```

```python
def load_qflow_ogbnech_results_csv(env_name, alpha, q_agg, discount, start_seed, end_seed, time_embed_dim, actor_loss_type="grad", metric_name="success"):
    root_dir = "/mnt/nas/jaehyeok/qflow/rebuttal/"
    results = []

    if time_embed_dim is not None:
        time_embed_str = f'te{time_embed_dim}'
    else:
        time_embed_str = ''
    tmp_env_name = env_name.replace("-v0", "").replace("-singletask-task", '-t')
    
    init_seed = 10001
    for seed in range(start_seed, end_seed + 1):
        # seed = init_seed + init_seed * seed
        exp_name = f"qflow-{tmp_env_name}-a{alpha}-qagg{q_agg}-{time_embed_str}-al{actor_loss_type}-sd{seed}"
        eval_results_dir = os.path.join(root_dir, "qflow", env_name, exp_name, "eval.csv")
        if os.path.exists(eval_results_dir):
            df = pd.read_csv(eval_results_dir)
            results.append(df[f'evaluation/{metric_name}'].values)
        else:
            print(f"No results found for {exp_name}")

    if not results:
        return np.array([])

    max_len = max(len(r) for r in results)
    completed = [r for r in results if len(r) == max_len]
    n_incomplete = len(results) - len(completed)
    if n_incomplete > 0:
        print(f"Skipping {n_incomplete} incomplete run(s) (expected {max_len} evals)")

    return np.array(completed)

hyperparams_dict = {
    "antmaze_large":{
        "alpha":1,
        "q_agg":"mean",
        "discount":0.99,
    },
    "antmaze_giant":{
        "alpha":0.2,
        "q_agg":"min",
        "discount":0.995,
    },
    "humanoid_medium":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.995,
    },
    "humanoid_large":{
        "alpha":1,
        "q_agg":"mean",
        "discount":0.995,
    },
    "antsoccer":{
        "alpha":0.5,
        "q_agg":"mean",
        "discount":0.995,
    },
    "cube_single":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.99,
    },
    "cube_double":{
        "alpha":2,
        "q_agg":"mean",
        "discount":0.99,
    },
    "puzzle_3":{
        "alpha":20,
        "q_agg":"min",
        "discount":0.99,
    },
    "puzzle_4":{
        "alpha":20,
        "q_agg":"min",
        "discount":0.99,
    },
    "scene":{
        "alpha":5,
        "q_agg":"mean",
        "discount":0.99,
    },   
}

all_results = []

for task_idx in range(1, 6):
    env_names = {
        "antmaze_large": f"antmaze-large-navigate-singletask-task{task_idx}-v0",
        "antmaze_giant": f"antmaze-giant-navigate-singletask-task{task_idx}-v0",
        "humanoid_medium": f"humanoidmaze-medium-navigate-singletask-task{task_idx}-v0",
        "humanoid_large": f"humanoidmaze-large-navigate-singletask-task{task_idx}-v0",
        "antsoccer": f"antsoccer-arena-navigate-singletask-task{task_idx}-v0",
        "cube_single": f"cube-single-play-singletask-task{task_idx}-v0",
        "cube_double": f"cube-double-play-singletask-task{task_idx}-v0",
        "puzzle_3": f"puzzle-3x3-play-singletask-task{task_idx}-v0",
        "puzzle_4": f"puzzle-4x4-play-singletask-task{task_idx}-v0",
        "scene": f"scene-play-singletask-task{task_idx}-v0",
    }
    env_key = 'antmaze_giant'
    env_name = env_names[env_key]

    results = load_qflow_ogbnech_results_csv(
        env_name=env_name,
        start_seed=0,
        end_seed=3,
        time_embed_dim=16,
        actor_loss_type='grad',
        **hyperparams_dict[env_key])

    all_results.append(results)  # each: (num_seeds, num_timesteps)

# Concatenate across tasks: (num_tasks * num_seeds, num_timesteps)
all_results = np.concatenate(all_results, axis=0).reshape(5, 4, -1)


# Final 3 averages
print(all_results.mean(axis=1).reshape(5, -1)[:, -3:].mean(axis=-1))

mean = all_results.mean(axis=0)
final3_per_seed = mean[:, -3:].mean(axis=0)
final1_per_seed = mean[:, -1]

print(f"Final 3 averages: {final3_per_seed.mean():.2f} ± {final3_per_seed.std():.2f}")
print(f"Final only average: {final1_per_seed.mean():.2f} ± {final1_per_seed.std():.2f}")
```
